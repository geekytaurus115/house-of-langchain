What’s happening?

The Google Vertex AI SDK (used by langchain under the hood) needs Application Default Credentials (ADC) to authenticate.

These credentials let your app access Google Cloud resources securely.

The error means it can’t find any credentials in your environment.

How to fix this step-by-step
1️⃣ Create a Service Account in Google Cloud Console

Go to your Google Cloud Console
.

Select your project or create a new one.

Navigate to IAM & Admin > Service Accounts.

Click Create Service Account.

Give it a name (e.g., vertex-ai-access).

Grant this service account the role: Vertex AI User (and possibly Viewer).

After creating, go to Keys tab and create a new key.

Choose JSON and download the key file (e.g., vertex-key.json).

2️⃣ Set the environment variable GOOGLE_APPLICATION_CREDENTIALS

This tells the Google SDK where to find your credentials JSON file.

On Linux/macOS terminal:

export GOOGLE_APPLICATION_CREDENTIALS="/path/to/vertex-key.json"


On Windows PowerShell:

setx GOOGLE_APPLICATION_CREDENTIALS "C:\path\to\vertex-key.json"


Make sure your Python environment can access this variable (restart terminal or IDE if needed).
